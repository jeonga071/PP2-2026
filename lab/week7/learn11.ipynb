## 내장함수
- abs(): 절대값 반환
- all(): 모든 항목이 참이면 True
- any(): 한 개의 항목이라도 참이면 True
- eval(): 수식의 구문 분석 및 실행
- sum(): 합계
- len(): 객체의 길이 계산
- list(): 리스트 생성
- map(): 리스트, 튜플 등의 항목에 주어진 함수를 적용한 후, 결과 반환
- dir(): 객체가 가지고 있는 변수나 함수 표시
- max(), min(): 크거나 작은 항목 반환
- enumerate(): 열거형 객체 반환
- filter(): 특정 조건을 만족하는 요소만 선정
- zip(): 2개의 시퀀스를 하나로 묶음


```python
def square(n):
    return n * n
lst = [1,2,3,4,5]
result = list(map(square, lst))
print(result)
```

```python
seasons = ["Spring", "Summer", "Fall", "Winter"]
list(enumerate(seasons, start = 0))
```

```python
def myfilter(x):
    return x > 3
result = filter(myfilter, (1,2,3,4,5,6))
print(list(result))
```

## 이터레이터
- for 루프와 함께 사용할 수 있는 객체
- _ _iter_ _(): 자신을 반환
- _ _next_ _(): 다음 반복을 위한 값 반환. 값이 없으면 StopIteration

```python
class MyCounter(object):
    def __init__(self, low, high):
        self.current = low
        self.high = high

    def __iter__(self):
        return self

    def __len__(self):
        return high-low+1

    def __next__(self):
        if self.current > self.high:
            raise StopIteration #raise: 오류 시 사용
        else:
            self.current += 1
            return self.current - 1

c = MyCounter(1, 10)
for i in c:
    print(i, end = ' ')
```

## 제너레이터
- yield 키워드를 사용하여 이터레이터를 생성

```python
def myGenerater():
    yield "first"
    yield "second"
    yield "third"

for word in myGenerater():
    print(word)
```

- add: 덧셈

- pow: 지수
- truediv: 실수 나눗셈
- floordiv: 정수 나눗셈
- mod: 나머지
- lt: 작음
- le: 작거나 같음
- eq: 같음
- ne: 같지 않음
- gt: 큼
- ge: 크거나 같음


```python
class Point:
    def __init__(self, x = 0, y = 0):
        self.x = x
        self.y = y
    def __add__(self, other):
        x = self.x + other.x
        y = self.y + other.y
        return Point(x, y)

    def __str__(self):
        return f"Point({self.x},{self.y})"

p1 = Point(1, 2)
p2 = Point(3, 4)
print(p1 + p2)
```
