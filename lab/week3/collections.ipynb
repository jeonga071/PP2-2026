```python
fruits = ("apple", "banana", "grape")
fruits += ("pear", "kiwi")
fruits  #튜플끼리 합치기 가능

fruits = "apple", "banana", "grape"
print(fruits)
type(fruits)    #괄호가 없어도 튜플로 인식
```

# set 실습
 - 추가, 삭제
 - 집합 연산

```python
fruits2 = {"apple", "banana", "grape"}
fruits2.add("kiwi") #추가
print(fruits2)
fruits2.remove("apple") #삭제
print(fruits2)
fruits2.clear() #모든 값 삭제
print(fruits2)

A = {"apple", "banana", "grape"}
B = {"apple", "banana", "kiwi"}
C = A | B   #union
print("합집합: ", C)

D = A & B   #intersection
print("교집합: ", D)

E = A - B   #difference
print("차집합: ", E)
```

# dictionary 실습
 - key:value가 쌍으로 된 원소들의 집합
 - 원소에 접근할 때 dict[key] 형태로 접근

```python
capitals = {"Korea":"Seoul", "USA":"Waashington", "UK":"London"}
print(capitals["Korea"])
#print(capitals["France"])   #해당 키가 존재하지 않음

capitals["France"] = "Paris"    #항목 추가

if "UK" in capitals:
    capitals.pop("UK")  #UK가 있을 경우 삭제

for key, values in captitals.item():
    print(key, ":", values)
print(capitals)
```
