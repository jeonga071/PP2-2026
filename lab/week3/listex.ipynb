## List 연습 노트 배열입니다.

리스트의 합병 연산 (+)
리스트의 곱 연산(*)
리스트 안의 값 존재 여부 확인(in, not in)
리스트 값 추가(insert)
리스트 깊은 복사 (copy)

```python
fruit1 = ["애플", "바나나"]
fruit2 = ["복숭아", "포도"]
print(fruit1 + fruit2)
print(fruit1 * 3)

if "애플" in fruit1:    #존재할 때
    print("애플이 있습니다.")
if "복숭아" not in fruit1:  # 없을 때
    print("복숭아가 없습니다.")

fruit1.insert(0, "딸기")    # 해당 자리에 생성, 기존 값은 뒤로 밀림
print(fruit1)
fruit1.remove("딸기")    # 해당 값 제거
print("딸기 삭제")

print(fruit1.index("애플")) # 값이 없으면 in 사용


# 깊은 복사
import copy
temps = [[1, 2], 3, 4]
svalues = copy.deepcopy(temps)

svalues[0][0] = 100
print(svalues)
print(temps)

```
