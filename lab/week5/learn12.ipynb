# 12장 상속
- 상속에 대한 이론과 실제 학습

```python
class Car:  #기본정보(부모 클래스)
    def __init__(self, maker, model, color, price):
        self.maker = maker
        self.model = model
        self.color = color
        self.price = price

    def setMaker(self, maker):
        self.maker = maker

    def getMaker(self): #self만 작성
        return self.maker

    def getDesc(self):
        return self.__str__()
    
    def __str__(self):
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"

class ElectricCar(Car):
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price) #부모와 연결시 호출 필수
        self.batterySize = bat_size

    def getBatterySize(self):
        return self.batterySize

    def setBatterySize(self, bat_size):
        self.batterySize = bat_size

    def __str__(self):
        return f"{super().__str__()}, Battery : {self.batterySize}kwh"

    def getDesc(self):
        return self.__str__()

class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad):
        super().__init__(maker, model, color, price)
        self.payLode = payLoad
    
    def __str__(self):
        return f"{super().__str__()}, payLoad: {self.payLode}kg"

car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.getDesc())
print(car1.model)

ecar1 = ElectricCar("Tesla", "Model S", "Red", 8000, 81)
print(ecar1.getDesc())
ecar1.setBatterySize(92)
print(ecar1.getDesc())

truck1 = Truck("Ford", "F-150", "Balck", 4000, 1000)
print(truck1)
print(truck1.getMaker())

print(type(truck1))
print(type(ecar1))
print(type(car1))

print(isinstance(truck1, Truck))
print(isinstance(ecar1, ElectricCar))
print(isinstance(car1, Car))

print(isinstance(truck1, Car))
print(isinstance(ecar1, Car))

#print(isinstance(truck1, vehicle))
```

# 속성변수의 가시성: public, protected(_로 시작) vate (_로 시작)

```python
class Parent(object):
    def __init__(self):
        self.__private_moeny = 1000
        self._protected_moeny = 500
        self.public_money = 100

    def __str__(self):
        return f"Perent has {self.__private_moeny} money"

class Child(Parent):
    def __init__(self):
        super().__init__()

    def __str__(self):
        return f"Child : {self._protected_moeny + self.public_money} public money"

def test():
    p = Parent()
    print(p)
    c = Child()
    print(c)

    print(p.public_money)
    print(p._protected_moeny)
    #print(p.__private_moeny)

    print(f"Child : {c.public_money}")
    print(f"Child : {c._protected_moeny}")
    #print(f"Child : {c.__private_moeny}")

test()
```

# 다중 상속: 2개 이상의 부모클래스로부터 상속받을 때

```python
class Base1:
    def __init__(self, v1 = 10):
        self.attr1 = v1

    def display1(self):
        print(f"Base1: {self.attr1}")

    def display(self):
        print(f"{self.attr1}")

class Base2:
    def __init__(self, v2 = 20):
        self.attr2 = v2

    def display2(self):
        print(f"Base2: {self.attr2}")
        
    def display(self):
        print(f"{self.attr2}")

class C(Base1, Base2):
    def __init__(self, v1, v2, v3):
        Base1.__init__(self, v1)
        Base2.__init__(self, v2)
        self.attr3 = v3

    def print(self):
        print(f"{self.attr1}, {self.attr2}, {self.attr3}")
        self.display1()
        self.display2()
        Base1.display(self)
        Base2.display(self)
        


def test():
    c = C(v1 = 1, v2 = 2, v3 = 3)
    c.print()

    c2 = C(4, 5, 6)
    c2.print()

if __name__ == "__main__":
    test()
```

### Overriding : 부모의 함수(매소드) 재정의

```python
import math

class Shape:
    def __init__(self, name = "도형"):
        self.name = name

    def draw(self):
        print(f"{self.name}을(를) 그립니다.")

    def get_area(self):
        print(f"Shape: {self.name}의 면적을 계산합니다.")

class Circle(Shape):
    def __init__(self, radius = 1, name = "원"):
        super().__init__(name)
        self.radius = radius

    def draw(self):
        print(f"Circle:{self.name}을(를) 계산합니다.")

    def get_area(self) -> float:
        print(f"Circle:{self.name}의 면적을 계산합니다.")
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width = 1, height = 1, name = "사각형"):
        super().__init__(name)
        self.width = width
        self.height = height

    def draw(self):
            print(f"Rectangle:{self.name}을(를) 계산합니다.")
    
    def get_area(self) -> float:
        print(f"Rectangle:{self.name}의 면적을 계산합니다.")
        return self.width * self.height

def test():
    c = Circle()
    c.draw()
    print(c.get_area())

    r1 = Rectangle(width = 5, height= 10)
    r1.draw()
    print(r1.get_area())

    print("-------------------------")
    circles = [Circle(1), Circle(2), Circle(3)]
    for c in circles:
        #c.draw()
        print(c.get_area())
    print("-------------------------")

    print("-------------------------")
    rectangle = [Rectangle(1, 1), Rectangle(2, 2), Rectangle(3, 3)]
    for r in rectangle:
        #r.draw()
        print(r.get_area())
    print("-------------------------")

    print("-------------------------")
    shapes = [Circle(1), Rectangle(1, 1), Circle(20), Rectangle(2, 3)]
    for s in shapes:
        print(s.get_area())
    print("-------------------------")

if __name__ == "__main__":
    test()
```

### 은행 함수 작성

```python
class Bank():
    def getInterestRate(self):
        return 0.0
class BadBank(Bank):
    def getInterestRate(self):
        return 10.0
class NormalBank(Bank):
    def getInterestRate(self):
        return 5.0
class GoodBank(Bank):
    def getInterestRate(self):
        return 3.0

def test():
    b1 = BadBank()
    b2 = NormalBank()
    b3 = GoodBank()
    print("BadBank의 이자율: " + str(b1.getInterestRate()))
    print("NormalBank의 이자율: " + str(b2.getInterestRate()))
    print("GoodBank의 이자율: " + str(b3.getInterestRate()))

if __name__ == "__main__":
    test()
```

# 다형성
- 많은(poly) + 모양(morph)의 의미. 주로 하나의 식별자로 다양한 클래스를 처리하는 것

```python
class Shape:
    def __init__(self, name):
        self.name = name
    def getArea(self):
        raise NotImplementedError("이것은 추상메소드입니다.")

class Circle(Shape):
    def __init__(self, name, radius):
        super().__init__(name)
        self.radius = radius

    def getArea(self):
        return 3.14 * self.radius ** 2
class Rectangle(Shape):
    def __init__(self, name, width, height):
        super().__init__(name)
        self.width = width
        self.height = height

    def getArea(self):
        return self.width * self.height

shapeList = [Circle("c1", 10), Rectangle("r1", 10, 10)]
for s in shapeList:
    print(s.getArea())
```

# 클래스 관계
* is-a 관계: 상속
    - 승용차는 차량의 일종이다(Car is a Vehicle)
    - 강아지는 동물의 일종이다(Dog is an animal) ...
* has-a 관계: 구성
    - 도서관은 책을 가지고 있다(Library has a book) ...

```python
class Card:
    suitNames = ["클럽", "다이아몬드", "하트", "스페이드"]
    rankNames = [None, "에이스", "2", "3", "4", "5", "6", "7", "8", "9", "10", "잭", "퀸", "킹"]

    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank

    def __str__(self):
        return Card.suitNames[self.suit] + " " + Card.rankNames[self.rank]

class Deck:
    def __init__(self):
        self.cards = []

        for suit in range(4):
            for rank in range(1, 14):
                card = Card(suit, rank)
                self.cards.append(card)
    def __str__(self):
        lst = [str(card) for card in self.cards]
        return str(lst)

if __name__ == "__main__":
    deck = Deck()
    print(deck)
```

# 가시성
- 파이썬은 접근 제어자가 엄격하게 적용되지 않음. 하지만 관례적으로 가시성 표현
    * public: 이름 앞에 밑줄 없음. (value)  - 어디서나 접근 가능
    * protected: 이름 앞에 밑줄 1개 (_value)  - 하위 클래스에서 접근 권장.
    * private: 이름 앞에 밑줄 2개 (__value)  - 클래스 내부에서만 접근.
    

# 추상 함수
- abc 모듈의 ABC와 @abstractmethod 데코레이터(Java)를 사용해 정의
- 반드시 하위 클래스에서 구현. 추상 메서드를 가진 클래스는 직접 인스턴스 생성 불가
- abc.ABC 클래스로 상속 받아야함.
- @abstractmethod 선언 및 정의 필요

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "멍멍"

d = Dog()
print(d.sound())
```
