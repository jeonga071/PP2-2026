### Lab: 매출 파일 처리
- 파일에 상점의 일별 매출이 저장되어 있다.
- 파일을 읽고 일별 평균 매출과 총 매출 계산 결과를 summary.txt에 저장
- sales.txt 생성

```python
infilename = input("입력 파일 이름: ")
outfilename = input("출력 파일 이름: ")

try:
    infile = open(infilename, "r", encoding="utf-8")
    outfile = open(outfilename, "w", encoding = "utf-8")

    sum = 0
    count = 0

    line = infile.readline()
    while line != "":
        s = int(line)
        sum += s
        count += 1
        line = infile.readline() #line에 다음 줄 읽기

    outfile.write("총매출 = " + str(sum) + "\n")
    outfile.write("평균 일매출 = " + str(sum / count))
except FileNotFoundError:
    print("파일을 찾을 수 없습니다.")
finally:
    infile.close()
    outfile.close()
```

### 문자 출현 횟수 계산 방법

```python
counter = [0] * 26
infile = open("mobydick.txt", "r")
ch = infile.read(1)
while ch != "":
    ch = ch.upper()
    if ch >= "A" and ch <= "Z":
        i = ord(ch) - ord("A")  #ord: 문자를 유니코드 값으로 변환
        counter[i] += 1
    ch = infile.read(1)
print(counter)
```

### 행맨 제작

```python
import random

guesses = ""
turns = 10

infile = open("words.txt", "r")
lines = infile.readlines()
word = random.choice(lines)

while turns > 0:
    failed = 0
    for char in word:
        if char in guesses:
            print(char, end = "")
        else:
            print("_", end="")
            failed += 1
    if failed == 1:
        print("사용자 승리")
        break
    print("")
    guess = input("단어를 추측하시오: ")
    guesses += guess
    if guess not in word:
        turns -= 1
        print("틀렸음!")
        print(str(turns) + "기회가 남았음!")
        if turns == 0:
            print("사용자 패배 정답은 " + word)
infile.close()
    
```
