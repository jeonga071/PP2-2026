### Lab: 매출 파일 처리
- 파일에 상점의 일별 매출이 저장되어 있다.
- 파일을 읽고 일별 평균 매출과 총 매출 계산 결과를 summary.txt에 저장
- sales.txt 생성

```python
infilename = input("입력 파일 이름: ")
outfilename = input("출력 파일 이름: ")

try:
    infile = open(infilename, "r", encoding="utf-8")
    outfile = open(outfilename, "w", encoding = "utf-8")

    sum = 0
    count = 0

    line = infile.readline()
    while line != "":
        s = int(line)
        sum += s
        count += 1
        line = infile.readline() #line에 다음 줄 읽기

    outfile.write("총매출 = " + str(sum) + "\n")
    outfile.write("평균 일매출 = " + str(sum / count))
except FileNotFoundError:
    print("파일을 찾을 수 없습니다.")
finally:
    infile.close()
    outfile.close()
```

### 문자 출현 횟수 계산 방법

```python
counter = [0] * 26
infile = open("mobydick.txt", "r")
ch = infile.read(1)
while ch != "":
    ch = ch.upper()
    if ch >= "A" and ch <= "Z":
        i = ord(ch) - ord("A")  #ord: 문자를 유니코드 값으로 변환
        counter[i] += 1
    ch = infile.read(1)
print(counter)
```

### 행맨 제작

```python
import random

guesses = ""
turns = 10

infile = open("words.txt", "r")
lines = infile.readlines()
word = random.choice(lines)

while turns > 0:
    failed = 0
    for char in word:
        if char in guesses:
            print(char, end = "")
        else:
            print("_", end="")
            failed += 1
    if failed == 1:
        print("사용자 승리")
        break
    print("")
    guess = input("단어를 추측하시오: ")
    guesses += guess
    if guess not in word:
        turns -= 1
        print("틀렸음!")
        print(str(turns) + "기회가 남았음!")
        if turns == 0:
            print("사용자 패배 정답은 " + word)
infile.close()
    
```

# 정규식
- 특정한 규칙을 가지고 있는 문자열들을 메타 문자를 이용하여 표현하는 수식
- re.search() 사용
    * ^ : 시작
    * $ : 끝
    * . : 문자
    * \d : 숫자
    * \w : 문자와 숫자
    * \s : 공백문자
    * \S : 공백문자 제외
    * *: 0회 이상 반복
    * +: 1회 이상 반복
    * [abc] : 문자 범위
    * [^abc]: 괄호 제외 문자 범위 

```python
import re

pw = input("패스워드를 입력하세요")
flag = 0
while True:
    if (len(pw) < 8):
        flag = -1
        break
    elif not re.search("[a-z]", pw):
        flag = -1
        break
    elif not re.search("[A-Z]", pw):
        flag = -1
        break
    elif not re.search("[0-9]", pw):
        flag = -1
        break
    elif not re.search("[_@#]", pw):
        flag = -1
        break
    else:
        flag = 0
        print("유효한 패스워드")
        break
if flag != 0:
    print("유효한 패스워드가 아닙니다.")
```

# 예외처리
- 오류의 종류
    * IOError: 파일을 열 수 없음
    * importError: 파이썬이 모듈을 찾을 수 없음
    * ValueError: 연산이나 내장 함수에서 인수가 적절하지 않은 값을 가짐
    * KeyboardInterrupt: 사용자가 인터럿트 키를 누름(Ctr-C/ Delete)
    * EOFError: 내장 함수가 파일의 끝을 만남
- Try-catch문: 예외가 있으면 일치되는 예외를 실행한 후 넘어감.

```python
try:
    f = open("test.txt", "w")
    f.write("테스트 파일 수정")
except IOError:
    print("파일을 찾을 수 없음")
finally:
    f.close()
```
