# 터틀그래픽 사용
- 정사각형 3개 그리기

```python
import turtle
t = turtle.Turtle()
t.shape("turtle")

def square(length: int) -> None:    #반환하지 않음
    t.down()
    for i in range(4) :
        t.forward(length)
        t.left(360/4)
    t.up()
    t.forward(50)

for i in range(3):
    square(100)

turtle.done()
```
