```python
class TV:
    def __init__(self, Channel = 1, Volumne = 5, On = True):
        self.Channel = Channel
        self.Volumne = Volumne
        self.On = On

    def turnOn(self):
        self.On = True
    def turnOff(self):
        self.On = False
    def changeChannel(self, Channel):
        self.Channel = Channel
    def changeVolume(self, Volumne):
        self.Volumne = Volumne

    def __str__(self):  #print(tv)처럼 안에 들어갈 때 출력
        if self.On:
            return f"Channel = {self.Channel}, Volumne = {self.Volumne}"
        else:
            return "전원이 꺼져있음"

mytv = TV()
mytv.turnOff()
```

```python
import math

class Circle:
    def __init__(self, radius = 0):
        self.radius = radius

    def getArea(self): # 그냥 사용하고 싶을 때
        print(math.pi * self.radius * self.radius)
    def getPerimeter(self):
        print(2 * math.pi * self.radius)

c = Circle(10)
c.getArea()
c.getPerimeter()
print(c.radius) #접근 가능
c.radius = 100 #변경 가능
c.getArea()
```
